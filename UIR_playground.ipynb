```python
from PIL import Image
import numpy as np
import os
import matplotlib.pyplot as plt

cols = 7
rows = 2
fig, axes = plt.subplots(rows, cols, figsize=(20, 8))
axes = axes.flatten()

for i in range(14):
    file = files[i]
    img = np.array(Image.open(os.path.join(dir1, file)).convert("RGB"), dtype=np.uint8)
    ax = axes[i]
    ax.imshow(img, interpolation='nearest', vmin=0, vmax=255)
    ax.set_title(f"Image {i + 1}")
    ax.axis("off")

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(rows, cols, figsize=(20, 8))
axes = axes.flatten()
for i in range(14):
    file = files[i]
    img = np.array(Image.open(os.path.join(dir2, file)).convert("RGB"), dtype=np.uint8)
    ax = axes[i]
    ax.imshow(img, interpolation='nearest', vmin=0, vmax=255)
    ax.set_title(f"Image {i + 1}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import numpy as np
import os
import matplotlib.pyplot as plt

cols = 7
rows = 4
fig, axes = plt.subplots(rows, cols, figsize=(30, 15))
plt.subplots_adjust(left=0, right=1, top=1, bottom=0,
                    wspace=0.02, hspace=0.02)
axes = axes.flatten()

for j in range(14):

    file = files[j]
    i = j
    if j>6:
        i=j+7

    img = np.array(Image.open(os.path.join(dir1, file)).convert("RGB"), dtype=np.uint8)
    ax = axes[i]
    ax.imshow(img, interpolation='nearest', vmin=0, vmax=255)
    ax.set_title(f"Image {i}")
    ax.axis("off")

    file = files[j]
    img2 = np.array(Image.open(os.path.join(dir2, file)).convert("RGB"), dtype=np.uint8)
    ax = axes[i +7]
    ax.imshow(img2, interpolation='nearest', vmin=0, vmax=255)
    ax.set_title(f"UIR {i}")
    ax.axis("off")

plt.tight_layout(pad=0)
plt.savefig(outpath,dpi=400)
plt.show()
```
